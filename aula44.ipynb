```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pickle
import cv2
```

```python
imagem = cv2.imread("workplace-1245776_1920.jpg")
```

```python
#cv2.imshow("grupo",imagem)
```

```python
detector_imagem = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
```

```python
imagem_cinza = cv2.cvtColor(imagem,cv2.COLOR_BGR2GRAY)
```

```python
deteccoes = detector_imagem.detectMultiScale(imagem_cinza,scaleFactor=1.3,minSize=(30,30))
```

```python
deteccoes
```

```python
for (x,y,l,a) in deteccoes:
    cv2.rectangle(imagem,(x,y),(x+l,y+a),(0,255,0),2)
cv2.imshow("grupo",imagem)
cv2.waitKey()
```

```python
imagem_corpo = cv2.imread("pessoas.jpg")
```

```python
detector_corpo = cv2.CascadeClassifier("fullbody.xml")
```

```python
imagem_cinza_corpo = cv2.cvtColor(imagem_corpo,cv2.COLOR_BGR2GRAY)
```

```python
deteccoes = detector_corpo.detectMultiScale(imagem_cinza_corpo,scaleFactor=1.1,minSize=(40,60))
```

```python
deteccoes
```

```python
for (x,y,l,a) in deteccoes:
    cv2.rectangle(imagem_corpo,(x,y),(x+l,y+a),(0,255,0),2)
cv2.imshow("corpo",imagem_corpo)
cv2.waitKey()
```
