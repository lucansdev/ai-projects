```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pickle
from apyori import apriori
```

```python
base_mercado1 = pd.read_csv("mercado.csv",header=None)
base_mercado1
```

```python
transacoes = []
for i in range(len(base_mercado1)):
    transacoes.append([str(base_mercado1.values[i,j]) for j in range(4)])
```

```python
transacoes
```

```python
regras = apriori(transacoes, min_support = 0.3, min_confidence = 0.8, min_lift = 2)
resultados = list(regras)
resultados
```

```python
A = []
B = []
suporte = []
confianca = []
lift = []

for resultado in resultados:
  #print(resultado)
  s = resultado[1]
  #print(s)
  result_rules = resultado[2]
  #print(result_rule)
  for result_rule in result_rules:
    #print(result_rule)
    a = list(result_rule[0])
    b = list(result_rule[1])
    c = result_rule[2]
    l = result_rule[3]
    #print(a, ' - ', b, ' - ', c, ' - ', l)
    A.append(a)
    B.append(b)
    suporte.append(s)
    confianca.append(c)
    lift.append(l)
```

```python
rules_df = pd.DataFrame({"A":A,"B":B,"suporte":suporte,"confiança":confianca,"lift":lift})
```

```python
rules_df.sort_values(by= "lift",ascending=False)
```

#BASE MERCADO 2

```python
base_mercado2 = pd.read_csv("mercado2.csv",header=None)
base_mercado2
```

```python
transacoes2 = []
for i in range(base_mercado2.shape[0]):
    transacoes2.append([str(base_mercado2.values[i,j]) for j in range(base_mercado2.shape[1])])
```

```python
regras2 = apriori(transacoes2,min_support = 0.003, min_confidence = 0.2, min_lift = 3)
resultados2 = list(regras2)
```

```python
A2 = []
B2 = []
suporte2 = []
confianca2 = []
lift2 = []

for resultado2 in resultados2:
  #print(resultado)
  s2 = resultado2[1]
  #print(s)
  result_rules2 = resultado2[2]
  #print(result_rule)
  for result_rule2 in result_rules2:
    #print(result_rule)
    a2 = list(result_rule2[0])
    b2 = list(result_rule2[1])
    c2 = result_rule2[2]
    l2 = result_rule2[3]
    #print(a, ' - ', b, ' - ', c, ' - ', l)
    A2.append(a2)
    B2.append(b2)
    suporte2.append(s2)
    confianca2.append(c2)
    lift2.append(l2)
```

```python
rules_df2 = pd.DataFrame({"A":A2,"B":B2,"suporte":suporte2,"confiança":confianca2,"lift":lift2})
```

```python
rules_df2.sort_values(by= "confiança",ascending=False)
```
