```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import pickle
```

```python
from sklearn import datasets

x_random,y_random = datasets.make_moons(n_samples=1500,noise= 0.09)
```

```python
grafico1 = px.scatter(x=x_random[:,0],y=x_random[:,1])
grafico1.show()
```

```python
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN
```

```python
Kmeans = KMeans(n_clusters=2)
rotulos_k = Kmeans.fit_predict(x_random)
grafico_k = px.scatter(x=x_random[:,0],y=x_random[:,1],color=rotulos_k)
grafico_k.show()
```

```python
hc = AgglomerativeClustering(n_clusters=2,linkage="ward")
rotulos_hc = hc.fit_predict(x_random)
grafico_hc = px.scatter(x=x_random[:,0],y=x_random[:,1],color=rotulos_hc)
grafico_hc.show()
```

```python
dbscan = DBSCAN(eps=0.1)
rotulos_db = dbscan.fit_predict(x_random)
grafico_db = px.scatter(x=x_random[:,0],y=x_random[:,1],color=rotulos_db)
grafico_db.show()
```
